## Exercise 1 - Download and read the IRIS dataset
The Iris dataset we have met it already a number of times.

https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data

Remember: the file has no header, so you need to specify it when you read it.
Since it has no header let's assign the following column names to it:
- sepal_length
- sepal_width
- petal_length
- petal_width
- label

```python
import pandas as pd

col_names = ["sepal_length", "sepal_width", "petal_length", "petal_width", "label"]

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

df = pd.read_csv(url, header=None, names=col_names)

print(df)
```

## Exercise 2 - Normalize feature by subtracting means and dividing by standard deviation

```python
grouped_df = df.groupby("label")

# sepal_length_mean = grouped_df["sepal_length"].mean()
# sepal_width_mean = grouped_df["sepal_width"].mean()
# petal_length_mean = grouped_df["petal_length"].mean()
# petal_width_mean = grouped_df["petal_width"].mean()

# sepal_length_std = grouped_df["sepal_length"].std()
# sepal_width_std = grouped_df["sepal_width"].std()
# petal_length_std = grouped_df["petal_length"].std()
# petal_width_std = grouped_df["petal_width"].std()

# 1. Seleziona le sole colonne numeriche delle feature (escludendo 'label')
features = df[["sepal_length", "sepal_width", "petal_length", "petal_width"]]

# 2. Calcola media e deviazione standard colonna per colonna (restituiscono una Series ciascuna)
mean_series = features.mean()
std_series = features.std()

# 3. Z-score normalization tramite broadcasting (riga 322 delle slide)
df_norm = (features - mean_series) / std_series
print(df_norm)


```
