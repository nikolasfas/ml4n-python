# Let's group IRIS data! 
In this notebook we will do again some analysis on IRIS but using grouping functions with Pandas

## Exercise 1 - Load the dataset
Download the dataset and read it directly with pandas into a dataframe with the pd.read_csv function directly from:
"https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

Remember: the file has no header, so you need to specify it when you read it.
Since it has no header let's assign the following column names to it:
- sepal_length
- sepal_width
- petal_length
- petal_width
- label

```python
import pandas as pd 

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

col_names = ["sepal_length", "sepal_width", "petal_length", "petal_width", "label"]

df = pd.read_csv(url, header=None, names=col_names)

print(df.head())
```

## Exercise 2 - compute the mean and std 
Compute the mean and std of each feature for each class by grouping the classes

```python
grouped_df = df.groupby("label")

keys = []
for key, group_df in grouped_df:
    keys.append(key)

print(keys)

sepal_length_mean = grouped_df["sepal_length"].mean()
sepal_width_mean = grouped_df["sepal_width"].mean()
petal_length_mean = grouped_df["petal_length"].mean()
petal_width_mean = grouped_df["petal_width"].mean()

print(f"The mean values for sepal length are:\n",sepal_length_mean)
print(f"The mean values for sepal width are:\n", sepal_width_mean)
print(f"The mean values for petal length are:\n",petal_length_mean)
print(f"The mean values for petal width are:\n", petal_width_mean)
```
