## Exercise 1 Array accessing 
Given an ndarray x = [[3.0, 2.0, 5.0], [1.0, 4.0, 2.0], [6.0, 7.0, 8.0]]
Return two arrays:
1) Compute the mean of the elements along the second column 
2) Retrieve the elements with values higher than 2 along the second and third rows 

```python
x = np.array([[3.0, 2.0, 5.0], [1.0, 4.0, 2.0], [6.0, 7.0, 8.0]])
secondColumn = x[:, 1:2]
meanSecondColumn = np.mean(secondColumn)

#print("Second column:", secondColumn)
print("Mean of second column:", meanSecondColumn)

second_thirdRows = x[1:3, :]
higherThanTwo = second_thirdRows > 2
selected = second_thirdRows[higherThanTwo]

print("Second and third rows:\n", selected)
```

## Exercise 2 Array accessing to update
Given the previous array x = [[3.0, 2.0, 5.0], [1.0, 4.0, 2.0], [6.0, 7.0, 8.0]]
1) set the values of the first column to 0, if they are greater than 2
2) then try to do the same for the first and third columns together. Does it work? 

```python
x = np.array([[3.0, 2.0, 5.0], [1.0, 4.0, 2.0], [6.0, 7.0, 8.0]])

# 1. Select the first column (index 0) as a 1D array: [3.0, 1.0, 6.0]
first_col = x[:, 0]

# 2. Boolean mask of which rows in column 0 exceed 2: [True, False, True]
mask = first_col > 2

# 3. Update only those matching rows, specifically in column 0:
x[mask, 0] = 0

print("Updated array:\n", x)

x1 = np.array([[3.0, 2.0, 5.0], [1.0, 4.0, 2.0], [6.0, 7.0, 8.0]])

# Create a boolean mask equal to x1 but with all False values
mask = np.zeros(x1.shape, dtype=bool)

# Set the elements of the mask to True where the first and third columns of x1 are greater than 2
mask[:, [0, 2]] = x1[:, [0, 2]] > 2

# Apply the mask to x1 and set the selected elements to 0
x1[mask] = 0

print("Updated array with mask:\n", x1)

```

## Exercise 3 Array concatenation and reshaping
Given two arrays x1 = [[3,1], [2,4]] and x2 = [[3, 4], [2,4], [2,5]]

Try to concatenate them vertically and then horizontally.
If something does not work try to reshape the arrays so that you can concatenate (or stack) them.

*Remember*: arrays can be concatenated if and only if all the dimensions match, with the **only exception of the concatenation dimension**.

```python
x1 = np.array([[3,1], [2,4]])
x2 = np.array([[3, 4], [2,4], [2,5]])

concatVert = np.concatenate((x1, x2))

print("Concatenated array vertically:\n", concatVert)

newx2 = x2.reshape(2, 3)

print("Reshaped x2:\n", newx2)

concatHoriz = np.concatenate((x1, newx2), axis=1)

print("Concatenated array horizontally:\n", concatHoriz)
```

## Exercise 4 Array splitting (Optional)

1) split it back into the two blocks it was made of, of 2 and 3 columns
2) split it into 2 parts of equal size along axis 0
3) try to split it into 2 parts of equal size along axis 1: what happens?

*Remember*: the split functions return a **list** of arrays. `hsplit` and `vsplit` are `split` with the axis already chosen.

```python
x = np.array([[3, 1, 3, 4, 2], [2, 4, 4, 2, 5]])

x1 = x[0, 0:4].reshape(2, 2)

x2 = x[[0], [4]]

equal = np.split(x, 2)[1]

x3 = np.append(equal, x2).reshape(3, 2)

print("Reshaped x:\n", x1)
print("Selected element from x:\n", x2)
print("Split and appended array:\n", x3)
```
